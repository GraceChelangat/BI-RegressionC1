# TRAIN / TEST SPLIT
### Member 2: Train and test split and feature selection

```python
TARGET = "daily_sales_revenue_kes"

# Remove the target and store ID from the predictors
X = df.drop(columns=[TARGET, "store_id"])
y = df[TARGET]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Dividing the target variable into quantile-based bins
y_bins = pd.qcut(y, q=5, labels=False, duplicates="drop")
print(y_bins.value_counts().sort_index())
```

```python
#performing the train/test split
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

#Split the data into 80% training and 20% testing sets to prevent data leakage
# 80/20 keeps 80% for model development and 20% for unbiased final testing.
# Since the target is a highly right-skewed continuous revenue target—the quantile bins help preserve its range across the 80/20 split.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_bins
    )
print("Training Set:", X_train.shape)
print("Testing Set:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
split_summary = pd.DataFrame({
    "Training": y_train.describe(),
    "Testing": y_test.describe()
})
display(split_summary)
```

```python
# Calculate key target statistics for both datasets
print("Training mean:", y_train.mean())
print("Testing mean:", y_test.mean())

print("Training median:", y_train.median())
print("Testing median:", y_test.median())

print("Training skewness:", y_train.skew())
print("Testing skewness:", y_test.skew())
```

```python
# Feature selection
# 1. Correlation-based feature selection
# Calculated on training data only to prevent test-data leakage.
# Select numerical predictors from the training data
numeric_train = X_train.select_dtypes(include=np.number)

train_correlations = numeric_train.corrwith(y_train)

# Sort correlations by absolute strength
train_correlations = train_correlations.sort_values(
    key=lambda x: x.abs(),
    ascending=False
)

# Display the correlations
display(train_correlations)
```

```python
#  bar chart of numerical feature correlations
plt.figure(figsize=(10, 6))

train_correlations.sort_values().plot(kind="barh")


plt.title("Training-Set Correlation with Daily Sales Revenue")
plt.xlabel("Pearson Correlation")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
# Preparing training data for mutual information analysis
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical columns in the training predictors
numeric_features = X_train.select_dtypes(include=np.number).columns.tolist()

# Identify categorical columns in the training predictors
categorical_features = X_train.select_dtypes(exclude=np.number).columns.tolist()

# Create an encoder for categorical variables
mi_preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "num",
            # Impute missing numeric values so mutual_info_regression doesn't choke on NaN
            SimpleImputer(strategy="median"),
            numeric_features
        )
    ]
)

# Fit the encoder using training data only
X_train_encoded = mi_preprocessor.fit_transform(X_train)

# Get the resulting feature names
mi_feature_names = mi_preprocessor.get_feature_names_out()

# Convert the encoded data into a DataFrame
X_train_mi = pd.DataFrame(
    X_train_encoded,
    columns=mi_feature_names,
    index=X_train.index
)

# Display the shape
print("Encoded training data shape:", X_train_mi.shape)
```

```python
#2. Feature selection based on mutual information
#Mutual Information detects broader/non-linear relationships.
from sklearn.feature_selection import mutual_info_regression

# Calculate mutual information using training data only
mi_scores = mutual_info_regression(
    X_train_mi,
    y_train,
    random_state=RANDOM_STATE
)

# Store the scores in a DataFrame
mi_results = pd.DataFrame({
    "feature": mi_feature_names,
    "mutual_information": mi_scores
})

# Sort features from highest to lowest score
mi_results = mi_results.sort_values(
    "mutual_information",
    ascending=False
)

# Display the strongest relationships
display(mi_results.head(15))
```

```python
# Select the 15 strongest features
top_mi = mi_results.head(15).sort_values("mutual_information")

# Create the plot
plt.figure(figsize=(10, 7))

plt.barh(
    top_mi["feature"],
    top_mi["mutual_information"]
)

# Add labels
plt.title("Top Features by Mutual Information")
plt.xlabel("Mutual Information")
plt.ylabel("Feature")

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Confirm that store_id is not included in the predictors
print("store_id in predictors:", "store_id" in X_train.columns)
```

```python
#Checking for highly correlated numerical predictors to reduce multicollinearity
# This helps avoid keeping features that provide almost duplicate information.

# Calculate correlations between numerical predictors using training data
predictor_corr = numeric_train.corr().abs()

# Keep only the upper triangle of the correlation matrix
upper_triangle = predictor_corr.where(
    np.triu(np.ones(predictor_corr.shape), k=1).astype(bool)
)

# Find predictor pairs with correlation above 0.80
high_corr_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

# Rename the columns
high_corr_pairs.columns = ["Feature 1", "Feature 2", "Absolute Correlation"]

# Keep only strong correlations
high_corr_pairs = high_corr_pairs[
    high_corr_pairs["Absolute Correlation"] > 0.80
]

# Sort from strongest to weakest
high_corr_pairs = high_corr_pairs.sort_values(
    "Absolute Correlation",
    ascending=False
)

# Display the results
display(high_corr_pairs)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Check whether online_order_count's missingness is linked to delivery_available
pd.crosstab(X_train["delivery_available"], X_train["online_order_count"].isna())
```

### Structural Missingness

`online_order_count` is missing when `delivery_available` is **"No"**.

This makes sense because a store that does not offer delivery will not have online orders. So the value is not missing because of an error it is simply **not applicable**.

This is called **structural missingness**.

```python
# Check whether avg_basket_size_kes missingness follows any pattern
display(pd.crosstab(X_train["delivery_available"], X_train["avg_basket_size_kes"].isna()))
display(pd.crosstab(X_train["store_format"], X_train["avg_basket_size_kes"].isna()))
display(pd.crosstab(X_train["store_size_category"], X_train["avg_basket_size_kes"].isna()))
```

### Random Missingness

About **5%** of the `avg_basket_size_kes` values are missing.

The missing values are not related to other variables such as delivery availability, store format or store size. There is no clear reason for the missing values.

Therefore this is considered **random missingness**.

```python
structural_missing_feature = ["online_order_count"]   # missing when delivery is "No"
random_missing_features    = ["avg_basket_size_kes"]   # missing with no clear pattern
other_numeric_features = [
    "store_age_years", "store_size_sqm", "num_staff",
    "promotions_active", "footfall_count",
    "loyalty_program_members", "local_events_nearby",
]
categorical_features = [
    "store_format", "location_type",
    "delivery_available", "store_size_category",
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        # Structural missingness -> impute with 0 (no delivery = 0 online orders)
        ("structural_num", Pipeline([
            ("impute_zero", SimpleImputer(strategy="constant", fill_value=0)),
            ("scale", StandardScaler()),
        ]), structural_missing_feature),

        # Random missingness -> impute with median
        ("random_num", Pipeline([
            ("impute_median", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), random_missing_features),

        # Complete numeric columns -> scale only
        ("num", Pipeline([
            ("impute_median", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), other_numeric_features),

        # Categorical columns
        ("cat", OneHotEncoder(handle_unknown="ignore", drop="first"),
         categorical_features),
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed  = preprocessor.transform(X_test)

print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
print("Any NaNs left in train?", np.isnan(X_train_processed).any())
print("Any NaNs left in test?", np.isnan(X_test_processed).any())
```

### Target Transformation

`daily_sales_revenue_kes` is always positive and has a strong right-skew (skewness ≈ **6.70**).

A `log1p` transformation is used to reduce the skew. After transformation the skewness decreases to about **0.18** which is much closer to a normal distribution.

`log1p` is used instead of Box-Cox because it is simpler and does not require estimating a λ parameter.

The model will work with the **log-transformed values**. After making predictions `expm1` is used to convert them back to **KES** for the final evaluation.

```python
y_train_log = np.log1p(y_train)
y_test_log  = np.log1p(y_test)

print("Skew before:", y_train.skew(), "| after log1p:", y_train_log.skew())
```

```python
recovered = np.expm1(y_train_log)
print("Max reconstruction error:", np.max(np.abs(recovered.values - y_train.values)))
```

MEMBER 4

Model Training and Cross-Validation

All models train on y_train_log (log-transformed daily_sales_revenue_kes,since the raw target was right-skewed).
Seven models are compared, chosen to span the bias-variance spectrum: Linear Regression and Ridge (high
bias, low variance - assume a straight-line relationship), Lasso and
ElasticNet (linear but with strong regularization), Decision Tree (low
bias, high variance on its own), and Random Forest / Gradient Boosting
(ensembles that reduce a tree's variance). Comparing across this range
shows whether the data favors a simple or a flexible model, rather than
assuming it.

Each model is evaluated using 5-fold cross-validation on the training set
only, wrapped in the same preprocessing pipeline from Member 3, and
results are shown as a chart so we can see how
consistent each model is across folds, not just its average score.

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(random_state=RANDOM_STATE),
    "Lasso": Lasso(random_state=RANDOM_STATE, max_iter=5000),
    "ElasticNet": ElasticNet(random_state=RANDOM_STATE, max_iter=5000),
    "Decision Tree": DecisionTreeRegressor(random_state=RANDOM_STATE, max_depth=6),
    "Random Forest": RandomForestRegressor(random_state=RANDOM_STATE, n_estimators=200, max_depth=8),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE, n_estimators=200, max_depth=3),
}

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_results = {}

print("Note: Lower RMSE is better (log scale).")
for name, model in models.items():
    pipeline = Pipeline(steps=[("preprocessor", preprocessor), ("model", model)])
    scores = cross_val_score(pipeline, X_train, y_train_log, cv=kfold,
                              scoring="neg_root_mean_squared_error", n_jobs=-1)
    cv_results[name] = -scores
    print(f"{name:20s} RMSE (log scale): {-scores.mean():.4f}  (+/- {scores.std():.4f})")

```

```python
#Boxplot
cv_results_df = pd.DataFrame(cv_results)
plt.figure(figsize=(10, 5))
sns.boxplot(data=cv_results_df, color="#4F6EA4")
plt.ylabel("Cross-Validated RMSE (log scale)")
plt.title("5-Fold Cross-Validation: RMSE Distribution by Model")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

**Bias-Variance Tradeoff: Learning Curves**

Rather than guessing at bias/variance from CV spread alone, this
 retrains each model at increasing training-set sizes and compares
 train error vs validation error at each size:
 train and validation error close together, both high -> underfitting (high bias)
 train error much lower than validation error (big gap) -> overfitting (high variance)

```python
from sklearn.model_selection import learning_curve

def bias_variance_panel(model_spec, ax):
    sizes, tr_scores, va_scores = learning_curve(
        estimator=model_spec["pipeline"],
        X=X_train,
        y=y_train_log,  # log scale, consistent with every other model here
        cv=kfold,
        scoring="neg_root_mean_squared_error",
        train_sizes=np.linspace(0.1, 1.0, 8),
        n_jobs=-1,
        shuffle=True,
        random_state=RANDOM_STATE,
    )
    tr_rmse = -tr_scores.mean(axis=1)
    va_rmse = -va_scores.mean(axis=1)

    sns.lineplot(x=sizes, y=tr_rmse, marker="o", ax=ax, label="Train", color="darkgreen")
    sns.lineplot(x=sizes, y=va_rmse, marker="s", ax=ax, label="Validation", color="firebrick")
    ax.set_xlabel("Rows used in training")
    ax.set_ylabel("RMSE (log scale)")
    ax.set_title(model_spec["label"])

# Three models picked to represent the spectrum: rigid, flexible, and in-between
model_specs = [
    {"label": "Linear Regression (simple, rigid)",
     "pipeline": Pipeline([("preprocessor", preprocessor), ("model", LinearRegression())])},
    {"label": "Decision Tree depth=12 (complex, flexible)",
     "pipeline": Pipeline([("preprocessor", preprocessor), ("model", DecisionTreeRegressor(max_depth=12, random_state=RANDOM_STATE))])},
    {"label": "Ridge (linear + shrinkage)",
     "pipeline": Pipeline([("preprocessor", preprocessor), ("model", Ridge(random_state=RANDOM_STATE))])},
]

fig, axs = plt.subplots(1, 3, figsize=(17, 4.5))
for spec, ax in zip(model_specs, axs):
    bias_variance_panel(spec, ax)
fig.suptitle("Learning curves: how each model responds to more training rows", y=1.02)
plt.tight_layout()
plt.show()
```

**Interpretation of Cross-Validation Results**

Linear Regression and Ridge achieve the lowest mean RMSE (0.33), so a
close-to-linear relationship explains most of the data at this sample
size (720 rows). Lasso and ElasticNet perform much worse (~0.56) because
their default regularization (alpha=1.0) is too strong, shrinking too
many coefficients toward zero - a candidate for tuning later, not a
failed approach. Random Forest and Gradient Boosting land in between,
slightly behind the linear models on average.

The learning curves confirm this directly rather than by proxy. Linear
Regression's train RMSE (0.32-0.33) and validation RMSE (0.33-0.42)
converge closely as training rows increase, meaning the model is
appropriately fit, not overfitting - it performs almost as well on data
it hasn't seen as on data it has. Decision Tree (depth=12), in contrast,
shows a train RMSE that stays near zero even with all 576 training rows
while its validation RMSE remains far higher (0.51-0.65) and never
closes the gap. This is the textbook signature of overfitting: the tree
is memorizing the training rows rather than learning a pattern that
generalizes. Ridge behaves almost identically to Linear Regression here,
confirming its regularization isn't doing much extra work at this sample
size and feature count.

Based on cross-validation and the learning curves together, Linear
Regression and Ridge are the strongest candidates going forward - both
accurate and clearly not overfitting
