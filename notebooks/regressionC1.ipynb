# TRAIN / TEST SPLIT
### Member 2: Train and test split and feature selection

```python
TARGET = "daily_sales_revenue_kes"

# Remove the target and store ID from the predictors
X = df.drop(columns=[TARGET, "store_id"])
y = df[TARGET]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Dividing the target variable into quantile-based bins
y_bins = pd.qcut(y, q=5, labels=False, duplicates="drop")
print(y_bins.value_counts().sort_index())
```

```python
#performing the train/test split
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

#Split the data into 80% training and 20% testing sets to prevent data leakage
# 80/20 keeps 80% for model development and 20% for unbiased final testing.
# Since the target is a highly right-skewed continuous revenue target—the quantile bins help preserve its range across the 80/20 split.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_bins
    )
print("Training Set:", X_train.shape)
print("Testing Set:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
split_summary = pd.DataFrame({
    "Training": y_train.describe(),
    "Testing": y_test.describe()
})
display(split_summary)
```

```python
# Calculate key target statistics for both datasets
print("Training mean:", y_train.mean())
print("Testing mean:", y_test.mean())

print("Training median:", y_train.median())
print("Testing median:", y_test.median())

print("Training skewness:", y_train.skew())
print("Testing skewness:", y_test.skew())
```

```python
# Feature selection
# 1. Correlation-based feature selection
# Calculated on training data only to prevent test-data leakage.
# Select numerical predictors from the training data
numeric_train = X_train.select_dtypes(include=np.number)

train_correlations = numeric_train.corrwith(y_train)

# Sort correlations by absolute strength
train_correlations = train_correlations.sort_values(
    key=lambda x: x.abs(),
    ascending=False
)

# Display the correlations
display(train_correlations)
```

```python
#  bar chart of numerical feature correlations
plt.figure(figsize=(10, 6))

train_correlations.sort_values().plot(kind="barh")


plt.title("Training-Set Correlation with Daily Sales Revenue")
plt.xlabel("Pearson Correlation")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
# Preparing training data for mutual information analysis
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical columns in the training predictors
numeric_features = X_train.select_dtypes(include=np.number).columns.tolist()

# Identify categorical columns in the training predictors
categorical_features = X_train.select_dtypes(exclude=np.number).columns.tolist()

# Create an encoder for categorical variables
mi_preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "num",
            # Impute missing numeric values so mutual_info_regression doesn't choke on NaN
            SimpleImputer(strategy="median"),
            numeric_features
        )
    ]
)

# Fit the encoder using training data only
X_train_encoded = mi_preprocessor.fit_transform(X_train)

# Get the resulting feature names
mi_feature_names = mi_preprocessor.get_feature_names_out()

# Convert the encoded data into a DataFrame
X_train_mi = pd.DataFrame(
    X_train_encoded,
    columns=mi_feature_names,
    index=X_train.index
)

# Display the shape
print("Encoded training data shape:", X_train_mi.shape)
```

```python
#2. Feature selection based on mutual information
#Mutual Information detects broader/non-linear relationships.
from sklearn.feature_selection import mutual_info_regression

# Calculate mutual information using training data only
mi_scores = mutual_info_regression(
    X_train_mi,
    y_train,
    random_state=RANDOM_STATE
)

# Store the scores in a DataFrame
mi_results = pd.DataFrame({
    "feature": mi_feature_names,
    "mutual_information": mi_scores
})

# Sort features from highest to lowest score
mi_results = mi_results.sort_values(
    "mutual_information",
    ascending=False
)

# Display the strongest relationships
display(mi_results.head(15))
```

```python
# Select the 15 strongest features
top_mi = mi_results.head(15).sort_values("mutual_information")

# Create the plot
plt.figure(figsize=(10, 7))

plt.barh(
    top_mi["feature"],
    top_mi["mutual_information"]
)

# Add labels
plt.title("Top Features by Mutual Information")
plt.xlabel("Mutual Information")
plt.ylabel("Feature")

# Display the plot
plt.tight_layout()
plt.show()
```

```python
#Checking for highly correlated numerical predictors to reduce multicollinearity
# This helps avoid keeping features that provide almost duplicate information.

# Calculate correlations between numerical predictors using training data
predictor_corr = numeric_train.corr().abs()

# Keep only the upper triangle of the correlation matrix
upper_triangle = predictor_corr.where(
    np.triu(np.ones(predictor_corr.shape), k=1).astype(bool)
)

# Find predictor pairs with correlation above 0.80
high_corr_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

# Rename the columns
high_corr_pairs.columns = ["Feature 1", "Feature 2", "Absolute Correlation"]

# Keep only strong correlations
high_corr_pairs = high_corr_pairs[
    high_corr_pairs["Absolute Correlation"] > 0.80
]

# Sort from strongest to weakest
high_corr_pairs = high_corr_pairs.sort_values(
    "Absolute Correlation",
    ascending=False
)

# Display the results
display(high_corr_pairs)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Check whether online_order_count's missingness is linked to delivery_available
pd.crosstab(X_train["delivery_available"], X_train["online_order_count"].isna())
```

### Structural Missingness

`online_order_count` is missing when `delivery_available` is **"No"**.

This makes sense because a store that does not offer delivery will not have online orders. So the value is not missing because of an error it is simply **not applicable**.

This is called **structural missingness**.

```python
# Check whether avg_basket_size_kes missingness follows any pattern
display(pd.crosstab(X_train["delivery_available"], X_train["avg_basket_size_kes"].isna()))
display(pd.crosstab(X_train["store_format"], X_train["avg_basket_size_kes"].isna()))
display(pd.crosstab(X_train["store_size_category"], X_train["avg_basket_size_kes"].isna()))
```

### Random Missingness

About **5%** of the `avg_basket_size_kes` values are missing.

The missing values are not related to other variables such as delivery availability, store format or store size. There is no clear reason for the missing values.

Therefore this is considered **random missingness**.

```python
structural_missing_feature = ["online_order_count"]   # missing when delivery is "No"
random_missing_features    = ["avg_basket_size_kes"]   # missing with no clear pattern
other_numeric_features = [
    "store_age_years", "store_size_sqm", "num_staff",
    "promotions_active", "footfall_count",
    "loyalty_program_members", "local_events_nearby",
]
categorical_features = [
    "store_format", "location_type",
    "delivery_available", "store_size_category",
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        # Structural missingness -> impute with 0 (no delivery = 0 online orders)
        ("structural_num", Pipeline([
            ("impute_zero", SimpleImputer(strategy="constant", fill_value=0)),
            ("scale", StandardScaler()),
        ]), structural_missing_feature),

        # Random missingness -> impute with median
        ("random_num", Pipeline([
            ("impute_median", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), random_missing_features),

        # Complete numeric columns -> scale only
        ("num", Pipeline([
            ("impute_median", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), other_numeric_features),

        # Categorical columns
        ("cat", OneHotEncoder(handle_unknown="ignore", drop="first"),
         categorical_features),
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed  = preprocessor.transform(X_test)

print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
print("Any NaNs left in train?", np.isnan(X_train_processed).any())
print("Any NaNs left in test?", np.isnan(X_test_processed).any())
```

```python
feature_names = preprocessor.get_feature_names_out()
list(feature_names)
```

### Target Transformation

`daily_sales_revenue_kes` is always positive and has a strong right-skew (skewness ≈ **6.70**).

A `log1p` transformation is used to reduce the skew. After transformation the skewness decreases to about **0.18** which is much closer to a normal distribution.

`log1p` is used instead of Box-Cox because it is simpler and does not require estimating a λ parameter.

The model will work with the **log-transformed values**. After making predictions `expm1` is used to convert them back to **KES** for the final evaluation.

```python
y_train_log = np.log1p(y_train)
y_test_log  = np.log1p(y_test)

print("Skew before:", y_train.skew(), "| after log1p:", y_train_log.skew())
```

```python
recovered = np.expm1(y_train_log)
print("Max reconstruction error:", np.max(np.abs(recovered.values - y_train.values)))
```

MEMBER 4

Model Training and Cross-Validation

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(random_state=RANDOM_STATE),
    "Lasso": Lasso(random_state=RANDOM_STATE, max_iter=5000),
    "ElasticNet": ElasticNet(random_state=RANDOM_STATE, max_iter=5000),
    "Decision Tree": DecisionTreeRegressor(random_state=RANDOM_STATE, max_depth=6),
    "Random Forest": RandomForestRegressor(random_state=RANDOM_STATE, n_estimators=200, max_depth=8),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE, n_estimators=200, max_depth=3),
}

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_results = {}

print("Note: Lower RMSE is better (log scale).")
for name, model in models.items():
    pipeline = Pipeline(steps=[("preprocessor", preprocessor), ("model", model)])
    scores = cross_val_score(pipeline, X_train, y_train_log, cv=kfold,
                              scoring="neg_root_mean_squared_error", n_jobs=-1)
    cv_results[name] = -scores
    print(f"{name:20s} RMSE (log scale): {-scores.mean():.4f}  (+/- {scores.std():.4f})")

```

```python
cv_results_df = pd.DataFrame(cv_results)
plt.figure(figsize=(10, 5))
sns.boxplot(data=cv_results_df, color="#4F6EA4")
plt.ylabel("Cross-Validated RMSE (log scale)")
plt.title("5-Fold Cross-Validation: RMSE Distribution by Model")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

### Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.linear_model import Ridge, Lasso, ElasticNet

# Placeholder model -- GridSearchCV swaps it out via param_grid below
tuning_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", Ridge())
])

param_grid = [
    {
        "model": [Ridge(random_state=RANDOM_STATE)],
        "model__alpha": [0.001, 0.01, 0.1, 1.0, 10.0, 100.0]
    },
    {
        "model": [Lasso(random_state=RANDOM_STATE, max_iter=20000)],
        "model__alpha": [1e-4, 1e-3, 1e-2, 1e-1, 1.0]
    },
    {
        "model": [ElasticNet(random_state=RANDOM_STATE, max_iter=20000)],
        "model__alpha": [1e-4, 1e-3, 1e-2, 1e-1, 1.0],
        "model__l1_ratio": [0.2, 0.5, 0.8]
    },
    {
        "model": [GradientBoostingRegressor(random_state=RANDOM_STATE)],
        "model__n_estimators": [100, 200, 300],
        "model__max_depth": [2, 3, 4],
        "model__learning_rate": [0.03, 0.1, 0.2],
        "model__subsample": [0.8, 1.0]
    }
]

grid_search = GridSearchCV(
    estimator=tuning_pipeline,
    param_grid=param_grid,
    cv=kfold,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    verbose=0
)

grid_search.fit(X_train, y_train_log)

print(f"Best tuned model: {grid_search.best_estimator_.named_steps['model']}")
print(f"Best parameters: {grid_search.best_params_}")
print(f"Best CV RMSE (log scale): {-grid_search.best_score_:.4f}")
```

### Quick Test-Set Check on the Winning Model

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

best_model = grid_search.best_estimator_
best_regressor = best_model.named_steps["model"]
best_model_name = best_regressor.__class__.__name__

y_pred_log = best_model.predict(X_test)
y_pred_kes = np.expm1(y_pred_log)
y_true_kes = y_test.values

rmse_kes = np.sqrt(mean_squared_error(y_true_kes, y_pred_kes))
mae_kes = mean_absolute_error(y_true_kes, y_pred_kes)
r2 = r2_score(y_true_kes, y_pred_kes)

print(f"Best tuned model: {best_model_name}")
print(f"Best params: {grid_search.best_params_}")
print(f"Best CV RMSE (log scale): {-grid_search.best_score_:.4f}")
print(f"Test MAE: {mae_kes:,.0f} KES | Test RMSE: {rmse_kes:,.0f} KES | Test R2: {r2:.3f}")

# Handed off to Member 5 for the full diagnostics and test-set evaluation write-up.
```

## Explainability

### Coefficient Interpretation

Because numeric features are standardized inside `preprocessor`, a linear model's coefficients are directly comparable in magnitude: a larger absolute coefficient indicates a stronger association with log-revenue, holding other features constant. If the winning model is tree-based instead, it has no `.coef_`, so we fall back to SHAP, PDP, and permutation importance below.

```python
feature_names_out = best_model.named_steps["preprocessor"].get_feature_names_out()

if hasattr(best_regressor, "coef_"):
    coefs_df = pd.DataFrame({
        "feature": feature_names_out,
        "coefficient": best_regressor.coef_
    }).sort_values("coefficient", key=abs, ascending=False)

    plt.figure(figsize=(9, 6))
    top_coefs = coefs_df.head(15)
    sns.barplot(data=top_coefs, y="feature", x="coefficient", color="#4F6EA4")
    plt.title(f"Top 15 Standardized Coefficients ({best_model_name})")
    plt.tight_layout()
    plt.show()
else:
    print(f"{best_model_name} does not provide .coef_.")
    print("Use permutation importance / SHAP / Partial Dependence Plots for explainability instead.")
```

### SHapley Additive exPlanations (SHAP) Values

SHAP explains individual predictions: for a given store-day, how much did each feature push the prediction up or down relative to the average prediction? It is built on Shapley values from cooperative game theory (Shapley, 1952), treating features as "players" sharing credit for the model's output.

- **Additive**: the SHAP values for one observation sum exactly to (prediction − baseline).
- **Local**: each observation gets its own explanation, not just one global summary.
- **Model-agnostic in principle**, though tree models use the faster `TreeExplainer` and linear models use `LinearExplainer`.

```python
import shap

X_test_sample = X_test.iloc[:500].copy()
X_test_transformed = best_model.named_steps["preprocessor"].transform(X_test_sample)

if "Forest" in best_model_name or "GradientBoosting" in best_model_name or "Tree" in best_model_name:
    explainer = shap.TreeExplainer(best_regressor)
    shap_values = explainer.shap_values(X_test_transformed)

    shap.summary_plot(shap_values, X_test_transformed, feature_names=feature_names_out, show=False)
    plt.title(f"SHAP Summary ({best_model_name})")
    plt.tight_layout()
    plt.show()

elif hasattr(best_regressor, "coef_"):
    masker = shap.maskers.Independent(X_test_transformed, max_samples=300)
    explainer = shap.LinearExplainer(best_regressor, masker=masker)
    shap_values = explainer(X_test_transformed)

    shap.summary_plot(shap_values.values, X_test_transformed, feature_names=feature_names_out, show=False)
    plt.title(f"SHAP Summary ({best_model_name})")
    plt.tight_layout()
    plt.show()

else:
    print(f"SHAP explainer not configured for model type: {best_model_name}")
```

### Partial Dependence Plots (PDP)

A PDP shows how the model's average prediction changes as one feature varies, holding all other features at their observed values and averaging across the dataset. We use `footfall_count` and `store_size_sqm`, the two numeric predictors with the strongest raw correlation to `daily_sales_revenue_kes`.

PDP is global and averaged (one curve for the whole dataset, unlike SHAP's per-observation view), and can mislead when features are correlated, since "holding everything else fixed" can generate unrealistic combinations.

```python
from sklearn.inspection import PartialDependenceDisplay

features_to_plot = ["footfall_count", "store_size_sqm"]

fig, ax = plt.subplots(figsize=(11, 4.5))
PartialDependenceDisplay.from_estimator(best_model, X_train, features=features_to_plot, ax=ax)
plt.suptitle(f"Partial Dependence ({best_model_name})")
plt.tight_layout()
plt.show()
```

### Permutation Feature Importance

Shuffle one feature's values (breaking its link to the target), re-score the model, and measure how much performance drops. A larger drop means the model relied more heavily on that feature. This works for any model type, not just tree-based ones, and is evaluated here on the held-out test set.

```python
from sklearn.inspection import permutation_importance

perm_result = permutation_importance(
    best_model,
    X_test,
    y_test_log,
    n_repeats=10,
    random_state=RANDOM_STATE,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

perm_importance_df = pd.DataFrame({
    "feature": X_test.columns,
    "importance_mean": perm_result.importances_mean,
    "importance_std": perm_result.importances_std,
}).sort_values("importance_mean", ascending=False)

plt.figure(figsize=(9, 6))
sns.barplot(
    data=perm_importance_df,
    y="feature",
    x="importance_mean",
    color="#4F6EA4",
    xerr=perm_importance_df["importance_std"]
)
plt.title(f"Permutation Feature Importance ({best_model_name}, test set)")
plt.xlabel("Mean decrease in RMSE score when feature is shuffled")
plt.tight_layout()
plt.show()
```

## Model Persistence

We persist the **entire fitted pipeline** (preprocessing + model) as a single object. Saving the model alone, without the exact fitted encoders/imputers/scaler, is a common error: new data at inference time must pass through identical preprocessing, and saving the steps separately risks a mismatch between training-time and inference-time transformations.

```python
import joblib
import os
from datetime import datetime

final_pipeline = grid_search.best_estimator_

# Refit on the full training data to be explicit (GridSearchCV already does this by default via refit=True)
final_pipeline.fit(X_train, y_train_log)

model_dir = "./model"
if not os.path.exists(model_dir):
    os.makedirs(model_dir)

model_name = final_pipeline.named_steps["model"].__class__.__name__
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
model_filename = f"{timestamp}_{model_name}_regressor_for_retail_sales.joblib".lower()
model_path = os.path.normpath(os.path.join(model_dir, model_filename))

joblib.dump(final_pipeline, model_path)
print("Selected model:", model_name)
print(f"Best pipeline (preprocessing + model) saved to: {model_path}")
```

```python
# Demonstrate loading the fitted pipeline and predicting on a new, unseen store
loaded_pipeline = joblib.load(model_path)

new_store = pd.DataFrame([{
    "store_format": "Supermarket",
    "location_type": "Mall",
    "delivery_available": "Yes",
    "store_size_category": "Medium",
    "store_age_years": 5.0,
    "store_size_sqm": 450.0,
    "num_staff": 12,
    "promotions_active": 1,
    "footfall_count": 800.0,
    "loyalty_program_members": 150,
    "online_order_count": 40.0,
    "avg_basket_size_kes": 1200.0,
    "local_events_nearby": 0,
}])

# The pipeline was trained on log1p(daily_sales_revenue_kes), so predictions come out
# on the log scale -- np.expm1 inverts this back to KES.
predicted_log_revenue = loaded_pipeline.predict(new_store)
predicted_revenue = np.expm1(predicted_log_revenue)[0]

print(f"Predicted daily sales revenue for new store: {predicted_revenue:,.0f} KES")
```
