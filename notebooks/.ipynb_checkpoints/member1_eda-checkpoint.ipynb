```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/retail_daily_sales.csv")
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

numeric_columns
```

```python
categorical_columns = df.select_dtypes(include="object")

categorical_columns
```

```python
for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["daily_sales_revenue_kes"],
    kde=True
)

plt.title("Distribution of Daily Sales Revenue")
plt.xlabel("Daily Sales Revenue (KES)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df["daily_sales_revenue_kes"]
)

plt.title("Boxplot of Daily Sales Revenue")
plt.xlabel("Daily Sales Revenue (KES)")

plt.show()
```
